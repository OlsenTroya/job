```python
import string
import pymorphy3
```

```python
morph = pymorphy3.MorphAnalyzer()
```

```python
def lemmatize_word(word):
    parsed = morph.parse(word)[0]
    return parsed.normal_form
```

```python
def lemmatize_text(text):
    words = text.split()
    lemmatized_words = [lemmatize_word(word) for word in words]
    return ' '.join(lemmatized_words)
```

```python
def remove_punctuation(text):
    translator = str.maketrans('', '', string.punctuation + '––')
    return text.translate(translator)
```

```python
def make_pairs(ind_words, n=1):
    for i in range(len(ind_words) - n):
        yield tuple(ind_words[i:i + n]), ind_words[i + n]
```

```python
def dictionary (text, n):
    words = text.split()
    pair = make_pairs(words, n)
    word_dictionary = {}
    for word_tuple, next_word in pair:
        if word_tuple in word_dictionary:
            if next_word in word_dictionary[word_tuple]:
                word_dictionary[word_tuple][next_word] += 1
            else:
                word_dictionary[word_tuple][next_word] = 1
        else:
          # Create a new entry for the word tuple with the next word count
            word_dictionary[word_tuple] = {next_word: 1}
    return word_dictionary
```

```python
def most_popularity_word(word_dictionary, word_tuple):
    if word_tuple in word_dictionary:
        next_words = word_dictionary[word_tuple]
        most_frequent_next_word = max(next_words, key=next_words.get)
        return most_frequent_next_word
    else:
        return None
```

```python
#data = open('cosmos.txt', encoding='utf8').read()
data = open('BlackHole.txt', encoding='utf8').read()
copy = remove_punctuation(data)
text = lemmatize_text(remove_punctuation(data))

initial_words = input(f"Enter words to search: ").lower()
n=len(initial_words.split())
if n==' ':
    print('You have not entered anything')


word_dictionary = dictionary (text, n)
lemmatized_input = ' '.join([lemmatize_word(word) for word in initial_words.split()])
word_tuple = tuple(lemmatized_input.split())
```

```python
while n != 0:
    next_word = most_popularity_word(word_dictionary, word_tuple)
    if next_word:
        next_word_index = text.split().index(next_word)
        print('Next word hint:', copy.split()[next_word_index])
        break
    else:
        n -= 1
        word_dictionary = dictionary (text, n)
        lemmatized_input = ' '.join([lemmatize_word(word) for word in initial_words.split()[:n]])
        word_tuple = tuple(lemmatized_input.split())
else:
    print('Nothing found')
```
